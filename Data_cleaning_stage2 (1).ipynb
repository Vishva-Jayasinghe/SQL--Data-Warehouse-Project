```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sales =pd.read_csv("sales_cleaned.csv")
products = pd.read_csv("products_cleaned.csv")
customers = pd.read_csv("customers_cleaned.csv")
```

```python

#drop column create_date
customers.drop(columns=["create_date"],inplace=True)
```

```python
customers.head()
```

```python
customers.insert(customers.columns.get_loc("customer_number") + 1,"full_name",customers["first_name"] + " " + customers["last_name"])
```

```python
customers.head()
```

```python
#drop the first_name and last_name
customers.drop(columns=["first_name","last_name"],inplace=True)
```

```python
customers.head()
```

```python
sales.head()
```

"The original dataset contained sales data from 2010–2014.
For realistic business analysis and forecasting demonstration, the dates were shifted to 2021–2025."

```python
sales["order_date"]= pd.to_datetime(sales["order_date"])
```

```python
date_cols = ["order_date", "shipping_date", "due_date"]
for col in date_cols:
    sales[col] = pd.to_datetime(sales[col]) + pd.DateOffset(years=10)
```

```python
sales.head()
```

Lets check the quality of the data frame after we changed.
